## Подготовка данных

### Импорты

```python
import pandas as pd
```

### Загрузка данных

```python
data = pd.read_excel('course_work_ml.xlsx')
pd.set_option('display.max_columns', None)
```

## Разведочный анализ данных

```python
df = data.copy()
```

```python
df = df.drop(columns=['Unnamed: 0'], axis = 1)
```

```python
df.info()
```

```python
null_columns = df.isnull().sum()
print(null_columns[null_columns > 0])

# Есть несколько колонок, у которых есть null значения
```

```python
df.duplicated().sum()
df.drop_duplicates(inplace=True)
# Удаляем строки с дубликатами
```

```python
df.describe()
```

```python
df.head()
```

## Преобразование данных

### Удаление столбцов

```python
stds = df.std(numeric_only=True)
uninformative_cols = stds[stds == 0].index

print(uninformative_cols)

# Удалим  дублирующиеся или неинформативные столбцы с нулевым стандартным отклонением
```

```python
df = df.drop(columns=uninformative_cols)
```

### Работа с null значениями

```python
numeric_cols = df.select_dtypes(include=['number']).columns
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].median())

# Заменяем null на медианное значение столбца
```

```python
null_columns = df.isnull().sum()
print(null_columns[null_columns > 0])

# null больше нет
```

### Работа с выбросами

```python
for col in ['IC50, mM', 'CC50, mM', 'SI']:
    if col in df.columns:
        q99 = df[col].quantile(0.99)
        df = df[df[col] <= q99]


# Удаляем строки, где значение "IC50, mM", "CC50, mM" и "SI" выше 99 перцентиля
```

## Выгрузка обработанного файла

```python
df.to_csv('processed_data.csv', index=False)
```
